# 4. Perform cluster analysis

# 4.1 Identify the right number of clusters

```python
# Step 1: preparing the clustering data
Cluster_Features = ['acousticness', 'danceability','energy','instrumentalness', 'liveness', 'loudness','speechiness','tempo','valence','popularity','duration_min']
X=df[Cluster_Features].copy()
```

```python
#checking missing values 
X.isnull().sum()
```

```python
# step 2 : Standardize the data 

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled=scaler.fit_transform(x)
```

```python
X_scaled[:5]
```

```python
# step 3 : Using the elbow method 
```

```python
from sklearn.cluster import KMeans
inertia=[]
for k in range(2,11):
    kmeans=KMeans(n_clusters=k, random_state=42,n_init=10)
    kmeans. fit(X_scaled)
    inertia. append(kmeans.inertia_)
```

```python
plt.figure(figsize=(10,6))
plt.plot(range(2,11),
         inertia,
         marker='o')
plt.xlabel('Number of clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal Number of Clusters')
plt.xticks(range(2,11))
plt.grid(True)
plt.show()
```

```python
#step 4 : using silhouette score

from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    
    score = silhouette_score(X_scaled, labels)
    
    silhouette_scores.append(score)
```

```python
plt.figure(figsize=(10,6))

plt.plot(range(2,11),
         silhouette_scores,
         marker='o')
plt.xlabel('numper of clusters(k)')
plt.ylabel('silhouette score')
plt.title('silhouette score for differenct number of clusters')
plt.xticks(range(2,11))
plt.grid(True)
plt.show()
```

```python
best_k = range(2, 11)[
    silhouette_scores.index(max(silhouette_scores))
]

print("Best K based on Silhouette Score:", best_k)
```

## 4.2 Apply K-Means Clustering

```python
optimal_k = 4
```

```python
kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)
```

```python
df['cluster'] = cluster_labels
```

## Step 6: Visualize the clusters using PCA

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)
```

```python
pca_df = pd.DataFrame(
    X_pca,
    columns=['PC1', 'PC2']
)

pca_df['cluster'] = cluster_labels
```

```python
plt.figure(figsize=(10, 7))

sns.scatterplot(
    data=pca_df,
    x='PC1',
    y='PC2',
    hue='cluster',
    palette='viridis',
    s=60
)

plt.title('Song Cohorts Using K-Means Clustering')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```
