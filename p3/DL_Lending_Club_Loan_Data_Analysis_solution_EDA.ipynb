Tasks:
  
1) **Feature Transformation** : Transform categorical values into numerical values (discrete)

which encoding method to be used for df.purpose columns ?

1) **Label encoding** - it is being used when we have relation between the values and it is not sequential.. E.g. : S , M , L , XL , XXL

2) **One Hot encoding** -  it is being used when we have no relation between the values E.g. - Names of Countries in a column,

```python
df_encoded = pd.get_dummies(df,columns=['purpose'],drop_first=True,dtype=int)  # drop_first -> its also a way to handle multi collinearity problem
df_encoded.head()
```

**Column name Treatment i.e. replace "." in column names with "_"**

```python
df_encoded.columns
```

**OPTION 1**

normally Used for renaming All the columns

```python
updates_col_names=[col.replace(".","_") for col in df_encoded.columns]
updates_col_names
```

```python
df_encoded.columns=updates_col_names
df_encoded.columns
```

**OPTION 2**

normally Used for renaming few columns

```python
for col in df_encoded.columns:

  # print(col)
  df_encoded.rename(columns={col:col.replace(".","_")},inplace=True)

df_encoded.columns
```
